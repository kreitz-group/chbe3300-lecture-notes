This notebook is the computational companion to the interlude on solving ODEs numerically in
[](#ch-reaction-kinetics). It integrates the constant-volume batch-reactor balance for a single
irreversible first-order reaction,

$$
\ce{A -> B}, \qquad \frac{\mathrm{d}c\un{A}}{\mathrm{d}t} = -k\, c\un{A},
$$

and compares the numerical solution against the analytical result
$c\un{A}(t) = c\un{A,0} \exp(-kt)$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

plt.rcParams["figure.dpi"] = 120
```

## Problem setup

Take a rate constant of $k = 0.35~\mathrm{min}^{-1}$ and an initial concentration of
$c\un{A,0} = 2.0~\mathrm{mol\,L}^{-1}$.

```python
k = 0.35  # 1/min
cA0 = 2.0  # mol/L
t_end = 15.0  # min


def rhs(t, y):
    """Right-hand side of the batch-reactor mass balance, dcA/dt = -k*cA."""
    (cA,) = y
    return [-k * cA]
```

## Integrate and compare

`solve_ivp` uses an adaptive Runge–Kutta method by default, which is far more accurate per step
than the explicit Euler–Cauchy method derived in the notes.

```python
sol = solve_ivp(rhs, [0.0, t_end], [cA0], dense_output=True, rtol=1e-8, atol=1e-10)

t = np.linspace(0.0, t_end, 200)
cA_numerical = sol.sol(t)[0]
cA_analytical = cA0 * np.exp(-k * t)

max_error = np.max(np.abs(cA_numerical - cA_analytical))
print(f"maximum absolute error: {max_error:.2e} mol/L")
```

```python
fig, ax = plt.subplots(figsize=(5, 3.2))
ax.plot(t, cA_analytical, label="analytical", lw=2)
ax.plot(t[::10], cA_numerical[::10], "o", ms=4, label="solve_ivp")
ax.set_xlabel("time / min")
ax.set_ylabel(r"$c_\mathrm{A}$ / (mol L$^{-1}$)")
ax.legend(frameon=False)
ax.set_xlim(0, t_end)
ax.set_ylim(0, None)
fig.tight_layout()
```
